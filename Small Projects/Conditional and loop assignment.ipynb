Q. W. A P. which takes one number from 0 to 9 from the user and prints 
it in the word. And if the word is not from 0 to 9 then 
it should print that number is outside of the range and program should 
exit. 
For exapmple:- 
input = 1 
output = one 

```python
# Dictionary to map digits to words
number_words = {
    0: 'zero',
    1: 'one',
    2: 'two',
    3: 'three',
    4: 'four',
    5: 'five',
    6: 'six',
    7: 'seven',
    8: 'eight',
    9: 'nine'
}
# Take input from user
try:
    user_input = int(input("Enter a number between 0 and 9: "))

    if user_input in number_words:
        print("Output:", number_words[user_input])
    else:
        print("The number is outside of the range.")
except ValueError:
    print("Invalid input! Please enter an integer between 0 and 9.")
```

Q. W. A P. to implement calculator but the operation to be done and two 
numbers will be taken as input from user:- 
Operation console should show below:- 
   Please select any one operation from below:- 
       * To add enter 1 
       * to subtract enter 2 
       * To multiply enter 3 
       * To divide enter 4 
       * To divide and find quotient enter 5 
       * To divide and find remainder enter 6 
       * To divide and find num1 to the power of num2 enter 7 
       * To Come out of the program enter 8 

```python
while True:
    # Operation menu
    print("\nPlease select any one operation from below:")
    print("  * To add enter 1")
    print("  * To subtract enter 2")
    print("  * To multiply enter 3")
    print("  * To divide enter 4")
    print("  * To divide and find quotient enter 5")
    print("  * To divide and find remainder enter 6")
    print("  * To find num1 to the power of num2 enter 7")
    print("  * To come out of the program enter 8")

    choice = input("Enter your choice (1-8): ")

    if choice == '8':
        print("Exiting the calculator. Goodbye!")
        break

    # For all other operations, take two numbers
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
    except ValueError:
        print("Invalid input! Please enter numeric values.")
        continue

    # Perform the selected operation
    if choice == '1':
        print("Result:", num1 + num2)
    elif choice == '2':
        print("Result:", num1 - num2)
    elif choice == '3':
        print("Result:", num1 * num2)
    elif choice == '4':
        if num2 == 0:
            print("Error: Cannot divide by zero.")
        else:
            print("Result:", num1 / num2)
    elif choice == '5':
        if num2 == 0:
            print("Error: Cannot divide by zero.")
        else:
            print("Quotient:", num1 // num2)
    elif choice == '6':
        if num2 == 0:
            print("Error: Cannot divide by zero.")
        else:
            print("Remainder:", num1 % num2)
    elif choice == '7':
        print("Result:", num1 ** num2)
    else:
        print("Invalid choice. Please enter a number between 1 and 8.")
```

Q. W A P which takes two numbers from the user and prints below output:- 
   1. num1 is greater than num2 if num1 is greater than num2 
   2. num1 is smaller than num2 if num1 is smaller than num2 
   3. num1 is equal to num2 if num1 and num2 are equal
Note:- 1. Do this problem using if - else 
     2. Do this using ternary operator 

```python
 # 1. Do this problem using if - else 
# Take two numbers as input
try:
    num1 = float(input("Enter first number (num1): "))
    num2 = float(input("Enter second number (num2): "))

    # Using if-else
    if num1 > num2:
        print("num1 is greater than num2")
    elif num1 < num2:
        print("num1 is smaller than num2")
    else:
        print("num1 is equal to num2")

except ValueError:
    print("Invalid input! Please enter numeric values.")

```

Q. Write a python program to find below output using loop:- 
 
Input:- 'peter piper picked a peck of pickled peppers.' 
Output:- ['peter', 'piper', 'picked', 'a', 'peck', 'of', 'pickled', 
'peppers']

```python
# Input string
my_str = 'peter piper picked a peck of pickled peppers.'

# Initialize empty list and word
words = []
word = ''

# Loop through each character in the string
for char in my_str:
    if char == ' ' or char == '.':
        if word:  # Add word only if it's not empty
            words.append(word)
            word = ''
    else:
        word += char

# In case the last word has no punctuation
if word:
    words.append(word)

# Output the result
print(words)
```

Q. Write a python program to find below output using loop:- 
 
Input:- 'peter piper picked a peck of pickled peppers.' 
Output:- 'retep repip dekcip a kcep fo delkcip sreppep'

```python
# Input string
text = 'peter piper picked a peck of pickled peppers.'

# Remove the period at the end temporarily
text = text.rstrip('.')

# Split into words
words = text.split()

# Initialize a list to hold reversed words
reversed_words = []

# Reverse each word using a loop
for word in words:
    reversed_word = ''
    for char in word:
        reversed_word = char + reversed_word  # reverse the word
    reversed_words.append(reversed_word)

# Join the reversed words into a sentence
result = ' '.join(reversed_words)

# Add the reversed period to the last word
result += '.'

# Print with single inverted commas
print("'" + result + "'")
```

Q. Write a python program to find below output using loop:- 
 
Input:- 'Peter Piper Picked A Peck Of Pickled Peppers.' 
Output:- 'Peter piper picked a peck of pickled peppers' 

```python
# Input string
text = 'Peter Piper Picked A Peck Of Pickled Peppers.'

# Remove the period at the end
text = text.rstrip('.')

# Split the sentence into words
words = text.split()

# Initialize an empty list to store the result words
formatted_words = []

# Loop through each word
for i in range(len(words)):
    word = words[i].lower()  # Convert entire word to lowercase
    if i == 0:
        word = word[0].upper() + word[1:]  # Capitalize only the first word
    formatted_words.append(word)

# Join the words into a sentence
result = ' '.join(formatted_words)

# Print with single inverted commas
print("'" + result + "'")
```

Q. Write a python program to implement index method using loop. If 
sub_str is found in my_str then it will print the index 
of first occurrence of first character of matching string in my_str:- 
 
Input:- my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.', 
sub_str = 'Pickl' 
Output:- 29 

```python
# Inputs
my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.'
sub_str = 'Pickl'

# Initialize variables
found = False

# Loop through my_str to check each possible starting index
for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    if match:
        print(i)  # Print the starting index of first match
        found = True
        break

# Optional: if not found, print not found message
if not found:
    print("Substring not found.")
```

Q. Write a python program to implement replace method using loop. If 
sub_str is found in my_str then it will replace the first 
occurrence of sub_str with new_str else it will will print sub_str not 
found:- 
 
Input:- my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.', 
sub_str = 'Peck', new_str = 'Pack' 
Output:- 'Peter Piper Picked A Pack Of Pickled Peppers.'

```python
# Inputs
my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.'
sub_str = 'Peck'
new_str = 'Pack'

# Initialize flag
found = False

# Loop to search for the first occurrence of sub_str
for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    if match:
        # Found the substring, construct new string manually
        result = my_str[:i] + new_str + my_str[i + len(sub_str):]
        print("'" + result + "'")
        found = True
        break

if not found:
    print("'" + sub_str + " not found'")
```

Q. Write a python program to find below output (implements rjust and 
ljust) using loop:- 
 
Input:- 'Peter Piper Picked A Peck Of Pickled Peppers.', sub_str = 
'Peck', 
Output:- '*********************Peck********************' 

```python
# Inputs
my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.'
sub_str = 'Peck'

# Calculate total width to match length of original string
total_width = len(my_str)
sub_len = len(sub_str)

# Calculate left and right padding
total_padding = total_width - sub_len
left_padding = total_padding // 2
right_padding = total_padding - left_padding  # In case of odd total_padding

# Create left padding using loop
left = ''
for _ in range(left_padding):
    left += '*'

# Create right padding using loop
right = ''
for _ in range(right_padding):
    right += '*'

# Combine all parts
centered_string = left + sub_str + right

# Print result with single inverted commas
print("'" + centered_string + "'")
```

Q. Write a python program to find below output using loop:- 
 
Input:- 'This is Python class', sep = ' is', 
Output:- ['This', 'is', 'Python class']

```python
# Inputs
my_str = 'This is Python class'
sep = ' is'

# Initialize result list and temp string
result = []
temp = ''
i = 0

while i < len(my_str):
    match = True
    # Check if the substring from current index matches the separator
    if i + len(sep) <= len(my_str):
        for j in range(len(sep)):
            if my_str[i + j] != sep[j]:
                match = False
                break
    else:
        match = False

    if match:
        result.append(temp)
        result.append(sep.strip())  # append 'is' without space
        temp = ''
        i += len(sep)
    else:
        temp += my_str[i]
        i += 1

# Append any remaining text
if temp:
    result.append(temp)

# Print result
print(result)
```
