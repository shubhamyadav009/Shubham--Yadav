```python
# Write a function to convert temperature from Celsius to Fahrenheit scale.
# oC to oF Conversion: Multipy by 9, then divide by 5, then add 32.

# Note: Return a string of 2 decimal places.
# In - Cel2Fah(28.0)
# Out - '82.40'
# In - Cel2Fah(0.00)
# Out - '32.00'

# Function to convert Celsius to Fahrenheit
def Cel2Fah(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return "{:.2f}".format(fahrenheit)

# Example usage
print(Cel2Fah(28.0))  # Output: '82.40'
print(Cel2Fah(0.00))  # Output: '32.00'

```

```python
# Write a function to compute the BMI of a person.
#     BMI = weight(kg)  /  ( height(m)*height(m) )

# Note: Return a string of 1 decimal place.
# In - BMI(63, 1.7)
# Out - '21.8'
# In - BMI(110, 2)
# Out - '27.5'


# Function to compute BMI
def BMI(weight, height):
    bmi = weight / (height * height)
    return "{:.1f}".format(bmi)

# Example usage
print(BMI(63, 1.7))   # Output: '21.8'
print(BMI(110, 2))    # Output: '27.5'
```

```python
# Write a function getSumOfLastDigits() that takes in a list of positive numbers and returns the sum of all the last digits in the list.
# getSumOfLastDigits([2, 3, 4])
# 9
# getSumOfLastDigits([1, 23, 456])
# 10

def getSumOfLastDigits(numbers):
    total = 0
    for num in numbers:
        last_digit = num % 10
        total += last_digit
    return total

# Example usage
print(getSumOfLastDigits([2, 3, 4]))       # Output: 9
print(getSumOfLastDigits([1, 23, 456]))    # Output: 10
```

```python
# Write a function that uses a default value.
# In - introduce('Lim', 20)
# Out - 'My name is Lim. I am 20 years old.'
# In - introduce('Ahmad')
# Out - 'My name is Ahmad. My age is secret.'

def introduce(name, age=None):
    if age is None:
        return f"My name is {name}. My age is secret."
    else:
        return f"My name is {name}. I am {age} years old."

# Example usage
print(introduce('Lim', 20))    # Output: My name is Lim. I am 20 years old.
print(introduce('Ahmad'))      # Output: My name is Ahmad. My age is secret.
```

```python
# Write a function isEquilateral(x, y, z) that accepts the 3 sides of a triangle as arguments. 
# The program should return True if it is an equilateral triangle.

# In - isEquilateral(2, 4, 3)
# False - False
# In - isEquilateral(3, 3, 3)
# Out - True
# In - isEquilateral(-3, -3, -3)
# Out - False

def isEquilateral(x, y, z):
    # All sides must be positive
    if x <= 0 or y <= 0 or z <= 0:
        return False
    # Check if all three sides are equal
    return x == y == z

# Example usage
print(isEquilateral(2, 4, 3))    # Output: False
print(isEquilateral(3, 3, 3))    # Output: True
print(isEquilateral(-3, -3, -3)) # Output: False
```

```python
# For a quadratic equation in the form of ax2+bx+c, the discriminant, D is b2-4ac. Write a function to compute the discriminant, D.
# In - quadratic(1, 2, 3)
# Out - 'The discriminant is -8.'
# In - quadratic(1, 3, 2)
# Out - 'The discriminant is 1.'
# In - quadratic(1, 4, 4)
# Out - 'The discriminant is 0.'


def quadratic(a, b, c):
    D = b**2 - 4*a*c
    return f"The discriminant is {D}."

# Example usage
print(quadratic(1, 2, 3))  # Output: The discriminant is -8.
print(quadratic(1, 3, 2))  # Output: The discriminant is 1.
print(quadratic(1, 4, 4))  # Output: The discriminant is 0.
```

```python
# Write a function find_max that accepts three numbers as arguments and returns the largest number among three. 
# Write another function main, in main() function accept three numbers from user and call find_max.

def find_max(a, b, c):
    return max(a, b, c)

def get_number(prompt):
    while True:
        user_input = input(prompt)
        try:
            number = float(user_input)
            return number
        except ValueError:
            print("Invalid input. Please enter a valid number.")

def main():
    a = get_number("Enter first number: ")
    b = get_number("Enter second number: ")
    c = get_number("Enter third number: ")
    
    largest = find_max(a, b, c)
    print("The largest number is:", largest)

# Call the main function
main()

```

```python
# Write a function, is_vowel that returns the value true if a given character is a vowel, 
# and otherwise returns false. 
# Write another function main, in main() function accept a string from user and count number 
# of vowels in that string.

def is_vowel(char):
    # Check if the character is a vowel (case-insensitive)
    return char.lower() in ['a', 'e', 'i', 'o', 'u']

def main():
    user_input = input("Enter a string: ")
    count = 0
    for ch in user_input:
        if is_vowel(ch):
            count += 1
    print("Number of vowels in the string:", count)

# Call main function
main()


```

```python
# Write a function named is_prime, which takes an integer as an argument and returns true 
# if the argument is a prime number, or false otherwise. 
# Also, write the main function that displays prime numbers between 1 to 500.

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

def main():
    print("Prime numbers between 1 and 500:")
    for num in range(1, 501):
        if is_prime(num):
            print(num, end=' ')
    print()  # for newline

# Call main function
main()
```

```python
# Write a function in python to find the sum of the cube of elements in a list. 
# The list is received as an argument to the function, in turn, the function must return the sum. 
# Write the main function which invokes the above function.

def sum_of_cubes(numbers):
    total = 0
    for num in numbers:
        total += num ** 3
    return total

def main():
    sample_list = [1, 2, 3, 4]
    result = sum_of_cubes(sample_list)
    print(f"The sum of cubes of {sample_list} is: {result}")

# Call the main function
main()

```

```python
# Write a method in python to display the elements of list thrice if it is a number and display 
# the element terminated with ‘#’ if it is not a number.

# For example, if the content of list is as follows :
# ThisList=[‘41’,‘DROND’,‘GIRIRAJ’, ‘13’,‘ZARA’]
# The output should be
# 414141
# DROND#
# GIRIRAJ#
# 131313
# ZARA#

def process_list(lst):
    for item in lst:
        if item.isdigit():  # Check if item is a number (all digits)
            print(item * 3)
        else:
            print(item + '#')

# Example usage
ThisList = ['41', 'DROND', 'GIRIRAJ', '13', 'ZARA']
process_list(ThisList)
```

```python
# For a given list of values in descending order, write a method in python to search 
# for a value with the help of Binary Search method. 
# The method should return position of the value and should return -1 if the value not present in the list.

def binary_search_desc(arr, target):
    low = 0
    high = len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid  # Return position (index) if found
        
        # Since list is descending, move accordingly
        if arr[mid] < target:
            high = mid - 1
        else:
            low = mid + 1
    
    return -1  # Not found

# Example usage
descending_list = [90, 80, 70, 60, 50, 40, 30, 20]
print(binary_search_desc(descending_list, 70))  # Output: 2
print(binary_search_desc(descending_list, 25))  # Output: -1
```

```python
# Write a function half_and_half that takes in a list and change the list 
# such that the elements of the second half are now in the first half.

# For example, if the size of list is even and content of list is as follows :
# my_liist = [10,20,30,40,50,60]
# The output should be
# [40,50,60,10,20,30]
# if the size of list is odd and content of list is as follows :
# my_liist = [10,20,30,40,50,60,70]
# The output should be
# [50,60,70,40,10,20,30]

def half_and_half(lst):
    n = len(lst)
    mid = n // 2
    if n % 2 == 0:
        # Even length: second half + first half
        lst[:] = lst[mid:] + lst[:mid]
    else:
        # Odd length: second half + middle element + first half (excluding middle)
        # The "middle" element is at index mid
        lst[:] = lst[mid+1:] + [lst[mid]] + lst[:mid]

# Example usage:
my_list_even = [10, 20, 30, 40, 50, 60]
half_and_half(my_list_even)
print(my_list_even)  # Output: [40, 50, 60, 10, 20, 30]

my_list_odd = [10, 20, 30, 40, 50, 60, 70]
half_and_half(my_list_odd)
print(my_list_odd)  # Output: [50, 60, 70, 40, 10, 20, 30]
```

```python
# Write a function that accepts a dictionary as an argument. If the dictionary contains replicate values, 
# return an empty dictionary, otherwise, return a new dictionary whose values are now the keys and whose 
# keys are the values.

def invert_dict_if_unique_values(d):
    # Check if there are duplicate values
    values = list(d.values())
    if len(values) != len(set(values)):
        return {}  # Duplicate values found, return empty dictionary
    
    # Invert keys and values
    inverted = {v: k for k, v in d.items()}
    return inverted

# Example usage:
d1 = {'a': 1, 'b': 2, 'c': 3}
print(invert_dict_if_unique_values(d1))  # Output: {1: 'a', 2: 'b', 3: 'c'}

d2 = {'a': 1, 'b': 2, 'c': 1}
print(invert_dict_if_unique_values(d2))  # Output: {}
```
