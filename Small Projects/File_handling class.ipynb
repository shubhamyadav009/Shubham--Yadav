```python
#global variable
lunchbox = 'Sandwich' 

def eat_lunch():
    #local variable
    drink = 'juice'
    print('inside function :')
    print('Lunch' , lunchbox)  # access the global variable here
    print('drink', drink)

eat_lunch()

print('outside function')
print('lunch', lunchbox)
print('drink', drink)
```
