Q. Arithmetic Operations on integers 
Take two different integer values. 
Store them in two different variables. 
Do below operations on them:- 
   Find sum of both numbers 
   Find difference between them 
   Find the product of both numbers. 
   Find value after dividing first num with second number 
   Find the remainder after dividing first number with second number 
   Find the quotient after dividing first number with second number 
   Find the result of the first num to the power of the second number. 

```python
# Input two integers
num1 = int(input("Enter first integer: "))
num2 = int(input("Enter second integer: "))

# Sum
sum_result = num1 + num2

# Difference
difference = num1 - num2

# Product
product = num1 * num2

# Division (float)
division = num1 / num2

# Remainder
remainder = num1 % num2

# Quotient (integer division)
quotient = num1 // num2

# Power
power = num1 ** num2

# Print results
print(f"Sum: {sum_result}")
print(f"Difference: {difference}")
print(f"Division: {division}")
print(f"Product: {product}")
print(f"Remainder: {remainder}")
print(f"Quotient: {quotient}")
print(f"Power: {power}")
```

Q. Comparison Operators on integers 
Take two different integer values. 
Store them in two different variables. 
Do below operations on them:- 
   Compare se two numbers with below operator:- 
       Greater than, '>' 
       Smaller than, '<' 
       Greater than or equal to, '>=' 
       Less than or equal to, '<=' 
Observe their output(return type should be boolean) 

```python
# Take input integers
num1 = int(input("Enter first integer: "))
num2 = int(input("Enter second integer: "))

# Greater than
gt = num1 > num2

# Smaller than
lt = num1 < num2

# Greater than or equal to
gte = num1 >= num2

# Less than or equal to
lte = num1 <= num2

# Print boolean results
print(f"{num1} > {num2} : {gt}")
print(f"{num1} < {num2} : {lt}")
print(f"{num1} >= {num2} : {gte}")
print(f"{num1} <= {num2} : {lte}")
```

Q. Bitwise Operators 
Do below operations on the values provided below:- 
   Bitwise and(&) -----------------------------------------> 10, 20 -------> Output is 0 
   Bitwise or(|)  -----------------------------------------> 10, 20 -------> Output is 30 
   Bitwise(^)     -----------------------------------------> 10, 20 -------> Output is 30 
   Bitwise negation(~) ------------------------------------> 10 -------> Output is -11 
   Bitwise left shift  ------------------------------------> 10,2 -------> Output is 40 
   Bitwise right shift ------------------------------------> 10,2 -------> Output is 2 
Cross check the output manually 

```python
# Given values
a = 10
b = 20

# Bitwise AND
and_result = a & b
print("Bitwise AND (&):", and_result)  # Output: 0

# Bitwise OR
or_result = a | b
print("Bitwise OR (|):", or_result)    # Output: 30

# Bitwise XOR
xor_result = a ^ b
print("Bitwise XOR (^):", xor_result)  # Output: 30

# Bitwise negation
neg_result = ~a
print("Bitwise Negation (~):", neg_result)  # Output: -11

# Bitwise left shift
left_shift_result = a << 2
print("Bitwise Left Shift (<< 2):", left_shift_result)  # Output: 40

# Bitwise right shift
right_shift_result = a >> 2
print("Bitwise Right Shift (>> 2):", right_shift_result)  # Output: 2
```

Q. Membership operation

print(10 in (10,10.20,10+20j,'Python')) 
print(2 in {1,2,3})

```python
print('2' in 'Python2.7.8')              # True, '2' is in the string
print(10 in [10, 10.20, 10+20j, 'Python'])  # True, 10 is in the list
print(10 in (10, 10.20, 10+20j, 'Python'))  # True, 10 is in the tuple
print(2 in {1, 2, 3})                   # True, 2 is in the set
print(3 in {1:100, 2:200, 3:300})      # True, 3 is a key in the dict
print(10 in range(20))                  # True, 10 is in the range 0 to 19
```

Q. An integer can be represented in binary, octal or hexadecimal form. 
Declare one binary, one octal and one hexadecimal value and store them 
in three different variables. 
Convert 9876 to its binary, octal and hexadecimal equivalent and print 
their corresponding value.

```python
# Declare values in different bases
binary_val = 0b1010        # binary for 10
octal_val = 0o12           # octal for 10
hexadecimal_val = 0xA      # hexadecimal for 10

print("Binary variable:", binary_val)
print("Octal variable:", octal_val)
print("Hexadecimal variable:", hexadecimal_val)

# Convert decimal 9876 to binary, octal and hexadecimal strings
num = 9876

binary_repr = bin(num)     # binary representation (string)
octal_repr = oct(num)      # octal representation (string)
hex_repr = hex(num)        # hexadecimal representation (string)

print("\nNumber 9876 in different bases:")
print("Binary:", binary_repr)
print("Octal:", octal_repr)
print("Hexadecimal:", hex_repr)
```
