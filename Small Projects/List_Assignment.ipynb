```python
#Write a Python program to find the sum of all elements in a list using loop.
#Input:- [10,20,30,40]
#Output:- 100

numbers = [10, 20, 30, 40]
total = 0

for num in numbers:
    total += num

print("Sum of all elements:", total)
```

```python
#Write a Python program to find the multiplication of all elements in a list using loop.
#Input:- [10,20,30,40]
#Output:- 240000

numbers = [10, 20, 30, 40]
product = 1

for num in numbers:
    product *= num

print("Multiplication of all elements:", product)
```

```python
#Write a Python program to find the largest number from a list using loop.
#Input:- [10,100,2321, 1,200,2]
#Output:- 2321

numbers = [10, 100, 2321, 1, 200, 2]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number in the list:", largest)

```

```python
#Write a Python program to count the number of strings having length more than 2 and are palindrome in a list using loop.
#Input:- ['ab', 'abc', 'aba', 'xyz', '1991']
#Output:- 2

words = ['ab', 'abc', 'aba', 'xyz', '1991']
count = 0

for word in words:
    if len(word) > 2 and word == word[::-1]:
        count += 1

print("Count of palindromic strings with length > 2:", count)
```

```python
#Write a Python program to sort a list in ascending order using loop.
#Input:- [100,10,1,298,65,483,49876,2,80,9,9213]
#Output:- [1,2,9,10,65,80,100,298,483,9213,49876]

numbers = [100, 10, 1, 298, 65, 483, 49876, 2, 80, 9, 9213]

# Using simple selection sort
for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]

print("Sorted list in ascending order:", numbers)



```

```python
#Write a Python program to get a sorted list in increasing order of last element in each tuple in a given list using loop.
#Input:- [(5,4),(9,1),(2,3),(5,9),(7,6),(5,5)]
#output:- [(9,1),(2,3),(5,4),(5,5),(7,6),(5,9)]

tuples_list = [(5, 4), (9, 1), (2, 3), (5, 9), (7, 6), (5, 5)]

# Using selection sort based on last element of tuple
for i in range(len(tuples_list)):
    for j in range(i + 1, len(tuples_list)):
        if tuples_list[i][-1] > tuples_list[j][-1]:
            tuples_list[i], tuples_list[j] = tuples_list[j], tuples_list[i]

print("Sorted list by last element in each tuple:", tuples_list)

```

```python
#Write a Python program to remove fuplicate element from a list using loop.
#Input:- [10,1,11,1,29,876,768,10,11,1,92,29,876]
#Output:- [10,1,11,29,876,768,92]

numbers = [10, 1, 11, 1, 29, 876, 768, 10, 11, 1, 92, 29, 876]
unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("List after removing duplicates:", unique)

```

```python
#Write a Python program to find the list of words that are longer than or equal to 4 from a given string.
#Input:- 'How much wood would a woodchuck chuck if a woodchuck could chuck wood'
#Output:- ['much', 'wood', 'would', 'woodchuck', 'chuck', 'could']
#Note:- Duplicate should be avoided.

input_str = 'How much wood would a woodchuck chuck if a woodchuck could chuck wood'
words = input_str.split()
result = []

for word in words:
    if len(word) >= 4 and word not in result:
        result.append(word)

print("Words longer than or equal to 4 characters:", result)

```

```python
#Write a Python program to create a 4X4 2D matrix with below elements using loop and list comprehension both.
#Output:- [[0,0,0,0],[0,1,2,3],[0,2,4,6],[0,3,6,9]]

inp_lst1 = [10, 20, 'Python', 10.20, 10+20j, [10, 20, 30], (10, 20, 30)]
inp_lst2 = [(10, 20, 30), 1, 20+3j, 100.2, 10+20j, [10, 20, 30], 'Python']

common_count = 0

for item1 in inp_lst1:
    for item2 in inp_lst2:
        if item1 == item2:
            common_count += 1
            break  # avoid counting duplicates multiple times

print(common_count >= 3)

```

```python
#Write a Python program which takes a list of numbers as input and prints a new list after removing even numbers from it.
#Input:- [10,21,22,98,87,45,33,1,2,100]
#Output:- [21,87,45,33,1]


numbers = [10, 21, 22, 98, 87, 45, 33, 1, 2, 100]
result = []

for num in numbers:
    if num % 2 != 0:
        result.append(num)

print(result)
```

```python
#Write a Python program which takes a list from the user and prints it after reshuffling the elements of the list.
#Input:- [10,21,22,98,87,45,33,1,2,100]
#Output:- [1,87,21,10,33,2,100,45,98,22] (It may be any randon list but with same elements)

import random

# Input list from user as comma separated values
user_input = input("Enter list elements separated by commas: ")
user_list = [int(x.strip()) for x in user_input.split(',')]

# Shuffle the list
random.shuffle(user_list)

# Print reshuffled list
print(user_list)
```
