```python
# In gene expression, mRNA is transcribed from a DNA template. 
# The 4 nucleotide bases of A, T, C, G corresponds to the U, A, G, C bases of the mRNA. 
# Write a function that returns the mRNA transcript given the sequence of a DNA strand.

# Use a dictionary to provide the mapping of DNA to RNA bases.

def dna_to_mrna(dna_seq):
    mapping = {'A': 'U', 'T': 'A', 'C': 'G', 'G': 'C'}
    mrna = ""

    for base in dna_seq:
        mrna += mapping.get(base.upper(), '')  # Handles uppercase DNA bases

    return mrna

# Example usage:
dna_sequence = "ATCGTAGC"
mrna_sequence = dna_to_mrna(dna_sequence)
print(mrna_sequence)
```

```python
# Write a function which takes a word as input and returns a dictionary with letters as key and no of time letters are repeated as value.
# In - count_letter('google.com')
# Out - {'g': 2, 'o': 3, 'l': 1, 'e': 1, '.': 1, 'c': 1, 'm': 1}

def count_letter(word):
    letter_count = {}
    for letter in word:
        letter_count[letter] = letter_count.get(letter, 0) + 1
    return letter_count

# Example usage:
result = count_letter('google.com')
print(result)
```

```python
# A DNA strand consisting of the 4 nucleotide bases is usually represented with a string of letters: A,T, C, G. 
# Write a function that computes the base composition of a given DNA sequence.

# In - baseComposition("CTATCGGCACCCTTTCAGCA")
# Out - {'A': 4, 'C': 8, 'T': 5,  'G': 3 }
    
# In - baseComposition("AGT")
# Out - {'A': 1, 'C': 0, 'T': 1,  'G': 1 }

def baseComposition(dna_seq):
    bases = ['A', 'C', 'T', 'G']
    composition = {base: 0 for base in bases}
    
    for base in dna_seq.upper():
        if base in composition:
            composition[base] += 1
    
    return composition

# Example usages:
print(baseComposition("CTATCGGCACCCTTTCAGCA"))
print(baseComposition("AGT"))
```

```python
# Write a function reverseLookup(dictionary, value) that takes in a dictionary 
# and a value as arguments and returns a sorted list of all keys that contains the value. 
# The function will return an empty list if no match is found.

# In - reverseLookup({'a':1, 'b':2, 'c':2}, 1)
# Out - ['a']
# In - reverseLookup({'a':1, 'b':2, 'c':2}, 2)
# Out - ['b', 'c']
# In - reverseLookup({'a':1, 'b':2, 'c':2}, 3)
# Out - []

def reverseLookup(dictionary, value):
    keys_with_value = [key for key, val in dictionary.items() if val == value]
    return sorted(keys_with_value)

# Example usages:
print(reverseLookup({'a':1, 'b':2, 'c':2}, 1))  # Output: ['a']
print(reverseLookup({'a':1, 'b':2, 'c':2}, 2))  # Output: ['b', 'c']
print(reverseLookup({'a':1, 'b':2, 'c':2}, 3))  # Output: []
```

```python
# Write a function invertDictionary(d) that takes in a dictionary as argument and return a dictionary that inverts the keys and the values of the original dictionary.
# In - invertDictionary({'a':1, 'b':2, 'c':3, 'd':2})
# Out - {1: ['a'], 2: ['b', 'd'], 3: ['c']}
# In - invertDictionary({'a':3, 'b':3, 'c':3})
# Out - {3: ['a', 'c', 'b']}
# In - invertDictionary({'a':2, 'b':1, 'c':2, 'd':1})
# Out - {1: ['b', 'd'], 2: ['a', 'c']}

def invertDictionary(d):
    inverted = {}
    for key, value in d.items():
        if value not in inverted:
            inverted[value] = []
        inverted[value].append(key)
    return inverted

# Example usages:
print(invertDictionary({'a':1, 'b':2, 'c':3, 'd':2}))  
# Output: {1: ['a'], 2: ['b', 'd'], 3: ['c']}

print(invertDictionary({'a':3, 'b':3, 'c':3}))  
# Output: {3: ['a', 'b', 'c']}  # Note: order may vary, but typically insertion order preserved in Python 3.7+

print(invertDictionary({'a':2, 'b':1, 'c':2, 'd':1}))  
# Output: {2: ['a', 'c'], 1: ['b', 'd']}

```

```python
# Write a function that converts a sparse vector into a dictionary as described above.
# In - convertVector([1, 0, 0, 2, 0, 0, 0, 3, 0, 0, 0, 0, 4])
# Out - {0: 1, 3: 2, 7: 3, 12: 4}
# In - convertVector([1, 0, 1 , 0, 2, 0, 1, 0, 0, 1, 0])
# Out - {0: 1, 2: 1, 4: 2, 6: 1, 9: 1}
# In - convertVector([0, 0, 0, 0, 0])
# Out - {}

def convertVector(lst):
    result = {}
    for index, value in enumerate(lst):
        if value != 0:
            result[index] = value
    return result

# Example usages:
print(convertVector([1, 0, 0, 2, 0, 0, 0, 3, 0, 0, 0, 0, 4]))
# Output: {0: 1, 3: 2, 7: 3, 12: 4}

print(convertVector([1, 0, 1 , 0, 2, 0, 1, 0, 0, 1, 0]))
# Output: {0: 1, 2: 1, 4: 2, 6: 1, 9: 1}

print(convertVector([0, 0, 0, 0, 0]))
# Output: {}
```

```python
# Write a function that converts a dictionary back to its sparse vector representation.
# In - convertDictionary({0: 1, 3: 2, 7: 3, 12: 4})
# Out - [1, 0, 0, 2, 0, 0, 0, 3, 0, 0, 0, 0, 4]
# In - convertDictionary({0: 1, 2: 1, 4: 2, 6: 1, 9: 1})
# Out - [1, 0, 1, 0, 2, 0, 1, 0, 0, 1]
# In - convertDictionary({})
# Out - []

def convertDictionary(d):
    if not d:
        return []
    max_index = max(d.keys())
    vector = [0] * (max_index + 1)
    for index, value in d.items():
        vector[index] = value
    return vector

# Example usages:
print(convertDictionary({0: 1, 3: 2, 7: 3, 12: 4}))
# Output: [1, 0, 0, 2, 0, 0, 0, 3, 0, 0, 0, 0, 4]

print(convertDictionary({0: 1, 2: 1, 4: 2, 6: 1, 9: 1}))
# Output: [1, 0, 1, 0, 2, 0, 1, 0, 0, 1]

print(convertDictionary({}))
# Output: []
```
