```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('Superstore.csv')

# Group by 'Category' and sum Sales and Profit
category_totals = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()

# Set figure size
plt.figure(figsize=(10, 6))

# Bar width and positions
bar_width = 0.35
x = range(len(category_totals))

# Create bar plots
plt.bar(x, category_totals['Sales'], width=bar_width, label='Sales', color='skyblue')
plt.bar([i + bar_width for i in x], category_totals['Profit'], width=bar_width, label='Profit', color='salmon')

# X-axis labels and ticks
plt.xticks([i + bar_width / 2 for i in x], category_totals['Category'])

# Labels and title
plt.xlabel('Category')
plt.ylabel('Amount')
plt.title('Sales and Profit by Category')
plt.legend()

# Layout adjustment
plt.tight_layout()

# Show plot
plt.show()
```

```python
import pandas as pd

# Load the data
df = pd.read_csv('Superstore.csv')

# Total sales in the dataset
total_sales = df['Sales'].sum()
print(f"Total Sales: {total_sales:.2f}")

# Group sales by customer and sort descending
customer_sales = df.groupby('Customer ID')['Sales'].sum().sort_values(ascending=False)

# Number of unique customers
num_customers = len(customer_sales)
print(f"Total unique customers: {num_customers}")

# Calculate the number of customers in the top 20%
top_20_count = int(0.2 * num_customers)
print(f"Top 20% customer count: {top_20_count}")

# Sum of sales from top 20% customers
top_sales = customer_sales.head(top_20_count).sum()
print(f"Top 20% Sales: {top_sales:.2f}")

# Calculate the percentage contribution of top 20%
percentage = 100 * top_sales / total_sales

# Final output
print(f"\nTop 20% customers contribute: ${top_sales:.2f}")
print(f"Percentage of total sales: {percentage:.2f}%")
```

```python
import pandas as pd

# Load data
df = pd.read_csv('Superstore.csv')

# Group by 'Product Name' (or the item column name) and sum 'Quantity'
item_sales = df.groupby('Product Name')['Quantity'].sum()

# Find the item with the highest quantity sold
most_sold_item = item_sales.idxmax()
most_sold_quantity = item_sales.max()

print(f"Most sold item: {most_sold_item}")
print(f"Quantity sold: {most_sold_quantity}")
```

```python
import pandas as pd

# Load data
df = pd.read_csv('Superstore.csv')

# Group by 'Product Name' and sum 'Profit'
item_profit = df.groupby('Product Name')['Profit'].sum()

# Find the item with the highest total profit
most_profitable_item = item_profit.idxmax()
highest_profit = item_profit.max()

print(f"Most profitable item: {most_profitable_item}")
print(f"Total profit: ${highest_profit:.2f}")
```
