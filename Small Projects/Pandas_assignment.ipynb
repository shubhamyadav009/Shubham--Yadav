```python
# Write a Python program to create and display a one
# dimensional array-like object containing an array of data using 
# Pandas module. 

import pandas as pd

# Create a one-dimensional array-like object (Series)
data = [10, 20, 30, 40, 50]
series = pd.Series(data)

# Display the Series
print(series)
```

```python
# Write a Python program to convert a Panda module Series to 
# Python list and its type. 

import pandas as pd

# Create a Pandas Series
series = pd.Series([10, 20, 30, 40, 50])

# Convert Series to Python list
list_from_series = series.tolist()

# Print the list and its type
print("List:", list_from_series)
print("Type of the converted object:", type(list_from_series))
```

```python
# Write a Python program to add, subtract, multiple and divide 
# two Pandas Series.

import pandas as pd

# Create two Pandas Series
series1 = pd.Series([10, 20, 30, 40])
series2 = pd.Series([1, 2, 3, 4])

# Element-wise operations
addition = series1 + series2
subtraction = series1 - series2
multiplication = series1 * series2
division = series1 / series2

print("Series 1:")
print(series1)

print("\nSeries 2:")
print(series2)

print("\nAddition:")
print(addition)

print("\nSubtraction:")
print(subtraction)

print("\nMultiplication:")
print(multiplication)

print("\nDivision:")
print(division)
```

```python
# Write a Python program to compare the elements of the two 
# Pandas Series.

import pandas as pd

# Create two Pandas Series
series1 = pd.Series([10, 20, 30, 40])
series2 = pd.Series([10, 25, 30, 35])

# Element-wise comparison
equal = series1 == series2
not_equal = series1 != series2
greater = series1 > series2
less = series1 < series2

print("Series 1:")
print(series1)

print("\nSeries 2:")
print(series2)

print("\nElement-wise Equal:")
print(equal)

print("\nElement-wise Not Equal:")
print(not_equal)

print("\nElement-wise Greater Than (Series1 > Series2):")
print(greater)

print("\nElement-wise Less Than (Series1 < Series2):")
print(less)
```

```python
# Write a Python program to display the following data column 
# wise.

import pandas as pd

# Sample data as a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

# Create DataFrame
df = pd.DataFrame(data)

# Display each column separately
for column in df.columns:
    print(f"Column: {column}")
    print(df[column])
    print()
```

```python
# Write a Python program to create and display a Data Frame 
# from a specified dictionary data which has the index labels. 

import pandas as pd

# Dictionary data
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

# Specify index labels
index_labels = ['a', 'b', 'c']

# Create DataFrame with custom index
df = pd.DataFrame(data, index=index_labels)

# Display the DataFrame
print(df)
```

```python
# Write a Python program to select the specified columns and 
# rows from a given Data Frame.

import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'Salary': [70000, 80000, 90000, 100000]
}

df = pd.DataFrame(data)

# Select specified columns: 'Name' and 'Salary'
columns_to_select = ['Name', 'Salary']

# Select specified rows by index (e.g., rows with index 1 and 3)
rows_to_select = [1, 3]

# Select rows and columns
selected_data = df.loc[rows_to_select, columns_to_select]

print("Original DataFrame:\n", df, "\n")
print("Selected rows and columns:\n", selected_data)
```

```python
# Write a Python program to select the rows the score is between 
# 15 and 20 (inclusive). 

import pandas as pd

# Sample DataFrame with 'score' column
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'score': [10, 15, 17, 20, 25]
}

df = pd.DataFrame(data)

# Select rows where score is between 15 and 20 inclusive
selected_rows = df[(df['score'] >= 15) & (df['score'] <= 20)]

print("Original DataFrame:\n", df, "\n")
print("Rows with score between 15 and 20 (inclusive):\n", selected_rows)
```

```python
# Write a Python program to select the rows where number of 
# attempts in the examination is less than 2 and score greater than 
# 15.

import pandas as pd

# Sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'attempts': [1, 2, 1, 3, 1],
    'score': [16, 14, 17, 20, 18]
}

df = pd.DataFrame(data)

# Select rows where attempts < 2 and score > 15
selected_rows = df[(df['attempts'] < 2) & (df['score'] > 15)]

print("Original DataFrame:\n", df, "\n")
print("Selected rows (attempts < 2 and score > 15):\n", selected_rows)
```

```python
# Write a Python program to append a new row 'k' to Data Frame 
# with given values for each column. Now delete the new row and 
# return the original data frame. 

import pandas as pd

# Original DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df, "\n")

# New row data as a dictionary
new_row = {'Name': 'David', 'Age': 40, 'City': 'Houston'}

# Append new row with index label 'k'
df.loc['k'] = new_row

print("DataFrame after appending new row 'k':\n", df, "\n")

# Delete the new row 'k'
df = df.drop('k')

print("DataFrame after deleting row 'k' (original restored):\n", df)
```

```python
# Write a Python program to sort the data frame first by 'name' in 
# descending order, then by 'score' in ascending order.

import pandas as pd

# Sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Bob', 'Alice'],
    'score': [85, 95, 78, 88, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df, "\n")

# Sort by 'name' descending, then by 'score' ascending
sorted_df = df.sort_values(by=['name', 'score'], ascending=[False, True])

print("Sorted DataFrame:\n", sorted_df)
```

```python
# Write a Python program to insert a new column in existing Data 
# Frame. 

import pandas as pd

# Existing DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df, "\n")

# Insert a new column 'City' with values
df['City'] = ['New York', 'Los Angeles', 'Chicago']

print("DataFrame after inserting new column 'City':\n", df)
```

```python
# Write a Python program to change the name 'James' to 'Suresh' 
# in name column of the data frame.

import pandas as pd

# Sample DataFrame
data = {
    'name': ['James', 'Bob', 'Charlie', 'James'],
    'age': [28, 34, 29, 42]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df, "\n")

# Replace 'James' with 'Suresh' in the 'name' column
df['name'] = df['name'].replace('James', 'Suresh')

print("DataFrame after changing 'James' to 'Suresh':\n", df)
```
