```python
# Write a function in python to count the number of lines 
# from a text file "story.txt" which is not starting with an alphabet "T". 

# Example: If the file "story.txt" contains the following lines: A boy is playing there.
# There is a playground.
# An aeroplane is in the sky.
# The sky is pink.
# Alphabets and numbers are allowed in the password.
# The function should display the output as 3

# Create story.txt with sample content
sample_text = """A boy is playing there.
There is a playground.
An aeroplane is in the sky.
The sky is pink.
Alphabets and numbers are allowed in the password."""

with open("story.txt", "w") as file:
    file.write(sample_text)

# Now count lines not starting with 'T'
def count_lines_not_starting_with_T(filename="story.txt"):
    count = 0
    with open(filename, "r") as file:
        for line in file:
            line = line.strip()
            if not line.startswith("T"):
                count += 1
    print(count)

count_lines_not_starting_with_T()

```

```python
# Write a function in Python to count and display the total number of words in a text file.

# Create a sample file named 'sample.txt'
with open('sample.txt', 'w') as f:
    f.write("Hello world!\n")
    f.write("This is a sample text file.\n")
    f.write("It has multiple lines and words.")

# Function to count words in a file
def count_words_in_file(filename):
    total_words = 0
    with open(filename, 'r') as file:
        for line in file:
            words = line.split()
            total_words += len(words)
    print(f"Total number of words in '{filename}': {total_words}")

# Call the function with the sample file
count_words_in_file('sample.txt')
```

```python
# Write a function in Python to read lines from a text file "notes.txt". 
# Your function should find and display the occurrence of the word "the".

# For example: If the content of the file is:
# "India is the fastest-growing economy. India is looking for more investments around the globe. The whole world is looking at India as a great market. Most of the Indians can foresee the heights that India is capable of reaching."

# The output should be 5.

def count_the_occurrences(filename="notes.txt"):
    count = 0
    with open(filename, "r") as file:
        for line in file:
            # Split the line into words and convert to lowercase for case-insensitive matching
            words = line.lower().split()
            # Count how many times 'the' appears in this line
            count += words.count("the")
    print(count)

# Example usage:
# First create the sample file with the provided content
sample_text = ("India is the fastest-growing economy. India is looking for more investments around the globe. "
               "The whole world is looking at India as a great market. Most of the Indians can foresee the heights "
               "that India is capable of reaching.")

with open("notes.txt", "w") as f:
    f.write(sample_text)

count_the_occurrences()
```

```python
# Write a function display_words() in python to read lines from a text file "story.txt", 
# and display those words, which are less than 4 characters.

def display_words(filename="story.txt"):
    with open(filename, "r") as file:
        for line in file:
            words = line.split()
            for word in words:
                # Remove punctuation around word (optional, but recommended)
                cleaned_word = word.strip(".,!?;:\"'()[]{}")
                if len(cleaned_word) < 4:
                    print(cleaned_word)

# Example usage:
# If story.txt doesn’t exist, you can create it like this:
sample_text = """A boy is playing there.
There is a playground.
An aeroplane is in the sky.
The sky is pink.
Alphabets and numbers are allowed in the password."""

with open("story.txt", "w") as f:
    f.write(sample_text)

display_words()
```

```python
# Write a function in Python to count the words "this" and "these" present in a text file "article.txt". 
# [Note that the words "this" and "these" are complete words] 

import re

def count_this_and_these(filename="article.txt"):
    count_this = 0
    count_these = 0
    pattern_this = re.compile(r'\bthis\b', re.IGNORECASE)
    pattern_these = re.compile(r'\bthese\b', re.IGNORECASE)

    with open(filename, "r") as file:
        for line in file:
            count_this += len(pattern_this.findall(line))
            count_these += len(pattern_these.findall(line))

    print(f"'this' occurs: {count_this} times")
    print(f"'these' occurs: {count_these} times")

# Example usage:
sample_text = """This is a test. These are examples.
this should be counted as well. These are important.
Is this the one? No, these are different."""

with open("article.txt", "w") as f:
    f.write(sample_text)

count_this_and_these()
```

```python
# Write a function in Python to count words in a text file those are ending with alphabet "e"

import re

def count_words_ending_with_e(filename):
    count = 0
    pattern = re.compile(r'\b\w*e\b', re.IGNORECASE)  # matches words ending with 'e'

    with open(filename, 'r') as file:
        for line in file:
            words = pattern.findall(line)
            count += len(words)
    print(f"Number of words ending with 'e': {count}")

# Example usage:
sample_text = """The apple is ripe. Please make sure to save the orange juice.
He wrote a note while the time was late."""

with open("example.txt", "w") as f:
    f.write(sample_text)

count_words_ending_with_e("example.txt")
```

```python
# Write a function in Python to count uppercase character in a text file.

def count_uppercase_chars(filename):
    count = 0
    with open(filename, 'r') as file:
        for line in file:
            for char in line:
                if char.isupper():
                    count += 1
    print(f"Number of uppercase characters: {count}")

# Example usage:
sample_text = """Hello World!
This is A Sample Text With Uppercase Letters."""

with open("textfile.txt", "w") as f:
    f.write(sample_text)

count_uppercase_chars("textfile.txt")
```

```python
# A text file named "matter.txt" contains some text, which needs to be displayed such that 
# every next character is separated by a symbol "#". 
# Write a function definition for hash_display() in Python that would display 
# the entire content of the file matter.txt in the desired format.

# Example :
# If the file matter.txt has the following content stored in it :
# THE WORLD IS ROUND

# The function hash_display() should display the following content :
# T#H#E# #W#O#R#L#D# #I#S# #R#O#U#N#D#

def hash_display(filename="matter.txt"):
    with open(filename, "r") as file:
        content = file.read()
        # Join every character with '#'
        output = '#'.join(content)
        print(output)

# Example usage: create the file with sample content first
sample_text = "THE WORLD IS ROUND"
with open("matter.txt", "w") as f:
    f.write(sample_text)

hash_display()
```

```python
# Aditi has used a text editing software to type some text. After saving the article as WORDS.TXT, 
# she realised that she has wrongly typed alphabet J in place of alphabet I everywhere in the article.
# Write a function definition for JTOI() in Python that would display the corrected version of entire content 
# of the file WORDS.TXT with all the alphabets "J" to be displayed as an alphabet "I" on screen.

# Note: Assuming that WORD.TXT does not contain any J alphabet otherwise.

# Example:
# If Aditi has stored the following content in the file WORDS.TXT:
# WELL, THJS JS A WORD BY JTSELF. YOU COULD STRETCH THJS TO BE A SENTENCE
# The function JTOI() should display the following content:
# WELL, THIS IS A WORD BY ITSELF. YOU COULD STRETCH THIS TO BE A SENTENCE

def JTOI(filename="WORDS.TXT"):
    with open(filename, "r") as file:
        content = file.read()
        corrected_content = content.replace('J', 'I')
        print(corrected_content)

# Example usage:
sample_text = "WELL, THJS JS A WORD BY JTSELF. YOU COULD STRETCH THJS TO BE A SENTENCE"

with open("WORDS.TXT", "w") as f:
    f.write(sample_text)

JTOI()
```

```python
# Write a function AMCount() in Python, which should read each character of a text file STORY.TXT, 
# should count and display the occurance of alphabets A and M (including small cases a and m too).
# For Example:
# If the file content is as follows:
# Updated information
# As simplified by official websites.
# The EUCount() function should display the output as:
# A or a:4
# M or m :2

def AMCount(filename="STORY.TXT"):
    count_a = 0
    count_m = 0
    with open(filename, "r") as file:
        for line in file:
            for char in line:
                if char == 'A' or char == 'a':
                    count_a += 1
                elif char == 'M' or char == 'm':
                    count_m += 1
    print(f"A or a: {count_a}")
    print(f"M or m: {count_m}")

# Example usage:
sample_text = """Updated information
As simplified by official websites."""

with open("STORY.TXT", "w") as f:
    f.write(sample_text)

AMCount()
```

```python
# A binary file "Book.dat" has structure [BookNo, Book_Name, Author, Price].
# i. Write a user defined function createFile() to input data for a record and add to Book.dat.
# ii. Write a function countRec(Author) in Python which accepts the Author name as parameter 
# and count and return number of books by the given Author are stored in the binary file "Book.dat"

import pickle
import os

def createFile():
    book_no = int(input("Enter Book Number: "))
    book_name = input("Enter Book Name: ")
    author = input("Enter Author Name: ")
    price = float(input("Enter Price: "))
    record = (book_no, book_name, author, price)
    with open("Book.dat", "ab") as file:
        pickle.dump(record, file)
    print("Record added successfully.\n")

def countRec(author_name):
    count = 0
    if not os.path.exists("Book.dat"):
        print("File does not exist.")
        return 0
    with open("Book.dat", "rb") as file:
        try:
            while True:
                record = pickle.load(file)
                if record[2].lower() == author_name.lower():
                    count += 1
        except EOFError:
            pass
    return count

def main():
    while True:
        print("1. Add book record")
        print("2. Count books by author")
        print("3. Exit")
        choice = input("Enter choice: ")
        if choice == '1':
            createFile()
        elif choice == '2':
            author = input("Enter author name to count books: ")
            result = countRec(author)
            print(f"Number of books by '{author}': {result}\n")
        elif choice == '3':
            print("Exiting program.")
            break
        else:
            print("Invalid choice. Try again.\n")

if __name__ == "__main__":
    main()

```

```python
# A binary file "STUDENT.DAT" has structure (admission_number, Name, Percentage). 
# Write a function count_rec() in Python that would read contents of the file "STUDENT.DAT" 
# and display the details of those students whose percentage is above 75. 
# Also display number of students scoring above 75%

import pickle
import os

def count_rec(filename="STUDENT.DAT"):
    if not os.path.exists(filename):
        print(f"{filename} does not exist.")
        return

    count = 0
    with open(filename, "rb") as file:
        try:
            while True:
                record = pickle.load(file)  # (admission_number, Name, Percentage)
                admission_number, name, percentage = record
                if percentage > 75:
                    print(f"Admission No: {admission_number}, Name: {name}, Percentage: {percentage}")
                    count += 1
        except EOFError:
            pass

    print(f"Number of students scoring above 75%: {count}")

# Example usage: Creating sample STUDENT.DAT with some records
sample_students = [
    (101, "Alice", 82.5),
    (102, "Bob", 74.0),
    (103, "Charlie", 91.2),
    (104, "Diana", 68.9),
    (105, "Ethan", 77.5)
]

with open("STUDENT.DAT", "wb") as f:
    for rec in sample_students:
        pickle.dump(rec, f)

count_rec()
```

```python
# Given a binary file employee.dat, created using dictionary object having keys: (empcode, name, and salary)

# Write a python function that add one more record at the end of file.
# Write a python function that display all employee records whose salary is more that 30000

import pickle
import os

def add_employee_record(filename, empcode, name, salary):
    new_record = {'empcode': empcode, 'name': name, 'salary': salary}

    if os.path.exists(filename):
        with open(filename, 'rb') as file:
            try:
                employee_list = pickle.load(file)
            except EOFError:
                employee_list = []
    else:
        employee_list = []

    employee_list.append(new_record)

    with open(filename, 'wb') as file:
        pickle.dump(employee_list, file)

    print(f"Added record: {new_record}")

def display_high_salary_employees(filename):
    if not os.path.exists(filename):
        print("File does not exist.")
        return

    with open(filename, 'rb') as file:
        try:
            employee_list = pickle.load(file)
            found = False
            print("Employees with salary > 30000:")
            for emp in employee_list:
                if emp['salary'] > 30000:
                    print(emp)
                    found = True
            if not found:
                print("No employees with salary > 30000.")
        except EOFError:
            print("File is empty.")

# Example usage
add_employee_record('employee.dat', 101, 'Alice', 35000)
add_employee_record('employee.dat', 102, 'Bob', 25000)
add_employee_record('employee.dat', 103, 'Charlie', 40000)

display_high_salary_employees('employee.dat')
```

```python
# A binary file school.dat has structure(rollno, name, class, fees)

# Write a definition for function total_fees( ) that reads each object of file and calculate 
# the total fees of students and display the same.

import pickle
import os

# Step 1: Create sample binary file (you only need to run this once)
def create_school_file():
    students = [
        {'rollno': 101, 'name': 'Alice', 'class': '10A', 'fees': 2000},
        {'rollno': 102, 'name': 'Bob', 'class': '10B', 'fees': 2500},
        {'rollno': 103, 'name': 'Charlie', 'class': '9C', 'fees': 1800},
    ]
    with open('school.dat', 'wb') as file:
        pickle.dump(students, file)
    print("school.dat created with sample student data.")

# Step 2: Define the function to calculate total fees
def total_fees(filename='school.dat'):
    if not os.path.exists(filename):
        print("File does not exist.")
        return

    with open(filename, 'rb') as file:
        try:
            student_list = pickle.load(file)
            total = sum(student['fees'] for student in student_list)
            print(f"Total fees collected from all students: ₹{total}")
        except EOFError:
            print("File is empty or corrupted.")

# Run example
create_school_file()
total_fees()
```

```python
# A binary file players.dat, containing records of following list format: [code, name, country and total runs]

# 1. Write a python function that display all records where player name starts from 'A'
# 2. Write a python function that accept country as an argument and count and display 
#     the number of players of that country.
# 3. Write a python function that add one record at the end of file.

import pickle
import os

# STEP 1: Create sample file (optional, run once)
def create_players_file():
    players = [
        [1, 'Alice', 'India', 5400],
        [2, 'Bob', 'Australia', 3200],
        [3, 'Andrew', 'England', 2800],
        [4, 'David', 'India', 6000],
        [5, 'Arnold', 'Australia', 4100],
    ]
    with open('players.dat', 'wb') as file:
        pickle.dump(players, file)
    print("players.dat created with sample data.\n")

# FUNCTION 1: Display all players whose name starts with 'A'
def display_players_starting_with_A(filename='players.dat'):
    print("Players whose name starts with 'A':")
    try:
        with open(filename, 'rb') as file:
            players = pickle.load(file)
            for player in players:
                if player[1].startswith('A'):
                    print(player)
    except FileNotFoundError:
        print("File not found.")
    except EOFError:
        print("File is empty.")

# FUNCTION 2: Count players from a given country
def count_players_by_country(country, filename='players.dat'):
    try:
        with open(filename, 'rb') as file:
            players = pickle.load(file)
            count = sum(1 for player in players if player[2].lower() == country.lower())
            print(f"Number of players from {country}: {count}")
    except FileNotFoundError:
        print("File not found.")
    except EOFError:
        print("File is empty.")

# FUNCTION 3: Add a new player record to the file
def add_player_record(code, name, country, total_runs, filename='players.dat'):
    new_player = [code, name, country, total_runs]

    players = []
    if os.path.exists(filename):
        with open(filename, 'rb') as file:
            try:
                players = pickle.load(file)
            except EOFError:
                pass  # empty file

    players.append(new_player)

    with open(filename, 'wb') as file:
        pickle.dump(players, file)

    print(f"Added new player record: {new_player}\n")

# --------------------------
# 🧪 TESTING THE FUNCTIONS
# --------------------------

# Run only once to create sample file
create_players_file()

# Display players whose name starts with 'A'
display_players_starting_with_A()

# Count players from 'India'
count_players_by_country('India')

# Add a new player
add_player_record(6, 'Anil', 'India', 7000)

# Re-check players starting with 'A'
display_players_starting_with_A()

# Re-count players from India
count_players_by_country('India')
```

```python
# Given a binary file game.dat, containing records of following list format: [game_name, participants]

# Write a function in Python that would read contents from the file game.dat and 
# creates a file named basket.dat copying only those records from game.dat where the game name is "Basket Ball"

import pickle

def create_sample_game_file():
    games = [
        ['Basket Ball', 10],
        ['Cricket', 22],
        ['Basket Ball', 8],
        ['Football', 11],
        ['Basket Ball', 12]
    ]
    with open('game.dat', 'wb') as file:
        pickle.dump(games, file)
    print("game.dat created with sample data.\n")

def copy_basketball_records():
    try:
        with open('game.dat', 'rb') as infile:
            games = pickle.load(infile)
            basket_games = [game for game in games if game[0] == 'Basket Ball']

        with open('basket.dat', 'wb') as outfile:
            pickle.dump(basket_games, outfile)

        print("basket.dat created with Basket Ball records.\n")

    except FileNotFoundError:
        print("game.dat not found.")
    except EOFError:
        print("game.dat is empty or corrupted.")
    except Exception as e:
        print("An unexpected error occurred:", e)

def display_basket_file():
    try:
        with open('basket.dat', 'rb') as file:
            data = pickle.load(file)
            print("Contents of basket.dat:")
            for record in data:
                print(record)
    except FileNotFoundError:
        print("basket.dat not found.")
    except EOFError:
        print("basket.dat is empty.")
    except Exception as e:
        print("An error occurred:", e)

# Run everything
create_sample_game_file()
copy_basketball_records()
display_basket_file()
```
